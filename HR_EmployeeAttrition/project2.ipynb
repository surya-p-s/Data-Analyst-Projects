## EDA

#### Univariant

```python
cat=[]
```

```python
for i in df:
    if(len(df[i].unique()) <=20 ):
        cat.append(i)
```

```python
df_cat=df[cat]
df_cat
```

```python
df_cat.drop(columns='EmployeeCount', inplace=True)
```

```python
cont=[]
```

```python
for i in df:
    if(len(df[i].unique())>=20):
        cont.append(i)
```

```python
df_cont=df[cont]
df_cont
```

```python
plotnumber =1
plt.figure(figsize=(20,25))
for i in df_cat:
    plt.subplot(7,4,plotnumber)
    sns.countplot(x=df_cat[i], hue=df_cat[i])
    plotnumber+=1
plt.tight_layout()
```

```python
plotnumber =1
plt.figure(figsize=(10,10))
for i in df_cont:
    plt.subplot(6,3,plotnumber)
    sns.histplot(df_cont[i], kde=True)
    plotnumber+=1
plt.tight_layout()
```

#### Bivarent Analysis

```python
plotnumber =1
plt.figure(figsize=(20,25))
for i in df_cat:
    plt.subplot(7,4,plotnumber)
    sns.countplot(x=df_cat[i], hue=df_cat['Gender'])
    plotnumber+=1
plt.tight_layout()
```

```python
plotnumber =1
plt.figure(figsize=(20,25))
for i in df_cat:
    plt.subplot(7,4,plotnumber)
    sns.countplot(x=df_cat[i], hue=df_cat['Department'])
    plotnumber+=1
plt.tight_layout()
```

```python
plotnumber =1
plt.figure(figsize=(20,25))
for i in df_cat:
    plt.subplot(7,4,plotnumber)
    sns.countplot(x=df_cat[i], hue=df_cat['OverTime'])
    plotnumber+=1
plt.tight_layout()
```
