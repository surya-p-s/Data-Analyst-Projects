### Data Cleaning

##### Handling with NULL Values

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df.Gender.value_counts()
```

```python
df.loc[df.Gender.isnull(),'Gender'] = 'Male'

# or
# df.Gender = df.Gender.fillna(value='Male')
```

```python
df.Gender.isnull().sum()
```

```python
df.Married = df.Married.fillna(value='Yes')
```

```python
df.Married.isnull().sum()
```

```python
df.Dependents = df.Dependents.fillna(value = '0')
```

```python
df.Dependents.isnull().sum()
```

```python
df.Self_Employed = df.Self_Employed.fillna(value='No')
```

```python
df.Self_Employed.isnull().sum()
```

```python
df.Loan_Amount_Term = df.Loan_Amount_Term.fillna(value = '360')
```

```python
df.Loan_Amount_Term.isnull().sum()
```

```python
df.isnull().sum()
```

```python
df.Credit_History.value_counts()
```

```python
df.loc[df.Credit_History.isnull(), 'Credit_History'] = 1.0
```

```python
df.Credit_History.isnull().sum()
```

```python
df.LoanAmount.value_counts()
```

```python
np.median(df.LoanAmount.dropna())
```

```python
df.loc[df.LoanAmount.isnull(), 'LoanAmount']=np.median(df.LoanAmount.dropna())
```

```python
df.LoanAmount.isnull().sum()
```
